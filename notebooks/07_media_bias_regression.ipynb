```python
df['clean_url'] = df['media_name'].str.replace(r'https?://(www\.)?', '', regex=True).str.rstrip('/')
dfm['clean_url'] = dfm['allsides_media_bias_ratings/publication/source_url'].str.replace(r'https?://(www\.)?', '', regex=True).str.rstrip('/')
```

```python
print('duplicate outlet URLs in AllSides table:', dfm.clean_url.duplicated().sum())  # >0 would duplicate headlines in the merge
merged_df = pd.merge(df, dfm, how='inner', left_on='clean_url', right_on='clean_url')
merged_df.shape 
```

```python
merged_df[["publish_date","title","allsides_media_bias_ratings/publication/media_bias_rating","clean_url"]]
```

```python
merged_df.rename(columns={"allsides_media_bias_ratings/publication/media_bias_rating": 'media_bias'}, inplace=True)
merged_df[["publish_date","title","media_bias","clean_url"]]
```

```python
merged_df["media_bias"].unique()
```

## Assigning Numerical Media Bias Ratings 

```python
response_key = {
    "Left" : -2,
    "Lean Left" : -1,
    "Center": 0,
    "Lean Right" : 1,
    "Right" : 2,    
}
```

```python
merged_df['bias_rating'] = merged_df['media_bias'].map(response_key)
```

```python
merged_df[["publish_date","title","media_bias","clean_url","bias_rating","pro_China","pro_Taiwan"]]
```

## Linear Regression



```python
md = merged_df
```

```python
merged_df["pro_Taiwan"].sum()
```

```python
import seaborn as sns
```

```python
sns.regplot(x = md.bias_rating, y = md.against_China , label = "Neg China (n={})".format(md.against_China.sum()),
            scatter=False, color = "salmon", line_kws={"linestyle":"-."})
sns.regplot(x = md.bias_rating, y = md.against_Taiwan , label = "Neg Taiwan",
            scatter=False, color = "cornflowerblue", line_kws={"linestyle":"--"})
sns.regplot(x = md.bias_rating, y = md.pro_China , label = "Pro China",
            scatter=False, color = "firebrick")
sns.regplot(x = md.bias_rating, y = md.pro_Taiwan , label = "Pro Taiwan",
            scatter=False, color = "navy", line_kws={"linestyle":":"})

plt.legend()
plt.xlabel("Bias Rating (Left to Right)")
plt.ylabel("Share of headlines (linear fit)")
plt.title("Media Coverage of Taiwan Across the Political Spectrum")
plt.savefig("media_bias.jpg", dpi=300)

# plt.xlim
```

## Performance Evaluation 

```python
import statsmodels.api as sm

# List of sentiment columns to model
sentiment_vars = ['pro_China', 'pro_Taiwan', 'against_China', 'against_Taiwan']

# Run linear regression for each sentiment variable
for var in sentiment_vars:
    X = sm.add_constant(md['bias_rating'])  # Add intercept
    y = md[var]
    
    model = sm.OLS(y, X).fit()
    
    print(f"--- Regression Results: {var} ---")
    print(f"R-squared: {model.rsquared:.3f}")
    print(f"Coef (bias_rating): {model.params['bias_rating']:.3f}")
    print(f"P-value: {model.pvalues['bias_rating']:.4f}")
    print(model.summary())
    print("\n")
```

## Logistic Regression

```python
feature_columns=["bias_rating"]
target_column="against_China"
```

```python
from sklearn.model_selection import train_test_split

taiwan_train, taiwan_test = train_test_split(merged_df, random_state=0) 
data_train = taiwan_train[feature_columns] 
data_test = taiwan_test[feature_columns] 

target_train = taiwan_train[target_column] 
target_test = taiwan_test[target_column]  
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

scaler = StandardScaler()
scaled_data_train = scaler.fit_transform(data_train)
scaled_data_test = scaler.transform(data_test)

# Now we'll fit the model
logistic_regression = LogisticRegression()
logistic_regression.fit(scaled_data_train, target_train)
accuracy = logistic_regression.score(scaled_data_test, target_test)
print(f"Accuracy on test set: {accuracy:.3f}")
```

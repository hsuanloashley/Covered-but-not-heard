```python
df = pd.concat([pd.read_csv(f) for f in 
                glob("../data/*.csv") ], ignore_index=True)
```

```python
a = pd.read_csv("../scraped_data/scraped_data.csv")
a = pd.concat((a, pd.read_csv("../scraped_data/scraped_data_2.csv")))
a = pd.concat((a, pd.read_csv("../scraped_data/scraped_data_3.csv")))
aa = a[~a.text.isna()]
success_ids = aa.sid.values
df2 = df[~df.id.isin(success_ids)]
df2 = df2.sample(frac=1, random_state=42)
```

```python
# Topic Modeling on Headlines.
# Use named entity recognition (NER) to identify key players.
# Come up with X number of themes you want to investigate, and understand their salience.

# vaderSentiment
```

```python
%%time
data = []
for n, (i, r) in enumerate(df2.iterrows()):
    try:
        url = r.url
        article = Article(url)
                
        # Download the article
        article.download()
        
        # Parse the article
        article.parse()
        
        # # NLP processing (optional, for more features like keywords, summary)
        # article.nlp()
        # Extracting required information
        article_data = {
            'sid' : r["id"],
            'title': article.title,
            'text': article.text,
            'publish_date': article.publish_date,
            'authors': article.authors,
            'keywords': article.keywords,
            # 'summary': article.summary,
            "media_url" : r.media_url,
            "url" : r.url
        }
    
        data.append( article_data )
    except Exception:
        article_data = {
            'sid' : r["id"],
            'title': np.nan,
            'text': np.nan,
            'publish_date': np.nan,
            'authors': np.nan,
            'keywords': np.nan,
            # 'summary': article.summary,
            "media_url" : r.media_url,
            "url" : r.url
        }
        data.append( article_data )
        
    if (n % 100) == 0:
        pd.DataFrame( data ).to_csv("../scraped_data/scraped_data_3.csv")
    # sleep(1)
```

```python
pd.DataFrame( data ).to_csv("../scraped_data/scraped_data_3.csv")
```
